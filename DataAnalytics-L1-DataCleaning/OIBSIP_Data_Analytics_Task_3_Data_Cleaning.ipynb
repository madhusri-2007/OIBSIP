```python
import pandas as pd

df = pd.read_csv("retail_sales_dataset.csv")

print("Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Unique Gender values:")
print(df['Gender'].unique())

print("\nUnique Product Category values:")
print(df['Product Category'].unique())
```

```python
numerical_columns = ['Age', 'Quantity', 'Price per Unit', 'Total Amount']

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    print(column, "-> Outliers:", len(outliers))
```

```python
df['Date'] = pd.to_datetime(df['Date'])

print(df.dtypes)
```

```python
summary = pd.DataFrame({
    'Column': df.columns,
    'Data Type': df.dtypes.astype(str),
    'Missing Values': df.isnull().sum(),
    'Unique Values': df.nunique()
})

summary
```

```python
df.to_csv("cleaned_retail_sales_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
